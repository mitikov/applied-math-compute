```python
import numpy as np
from scipy.optimize import fsolve
from ipywidgets import interact, FloatText
```

```python
def f(x):
    return np.sqrt(x + 1) - 1 / x

def check(start=1.0):
    roots, details, status, message = fsolve(f, start, full_output=True)
    root = float(roots[0])
    print("Кандидат:", root)
    print("Нев'язка:", abs(float(f(root))))
    print("Успішне завершення:", status == 1)
    print(message)

interact(check, start=FloatText(value=1.0, description="Старт"))
```

```python
import sympy as sp
from ipywidgets import interact, FloatText

x = sp.symbols('x')
# Ваша функція нев'язки (з використанням sp.sqrt замість np.sqrt)
formula = sp.sqrt(x + 1) - 1 / x

def check_again(start=1.0):
    try:
        root = sp.nsolve(formula, x, start)
        print("Кандидат nsolve:", root)
        print("Нев'язка:", abs(formula.subs(x, root).evalf()))
    except (ValueError, ZeroDivisionError) as error:
        print("Пошук не завершився:", error)

interact(check_again, start=FloatText(value=0.8, description="Старт"))
```
